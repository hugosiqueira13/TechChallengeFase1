# 01 — Análise Exploratória

Este notebook explora o dataset Iris e valida o pipeline de pré-processamento.

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
df = iris.data.copy()
df['target'] = iris.target.map(dict(enumerate(iris.target_names)))
df.head()
```

```python
print(df.describe())
print('\nValores nulos:\n', df.isnull().sum())
```

```python
sns.pairplot(df, hue='target', diag_kind='kde')
plt.suptitle('Iris — Pairplot', y=1.02)
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for ax, col in zip(axes, iris.data.columns):
    for species in df['target'].unique():
        df[df['target'] == species][col].hist(ax=ax, label=species, alpha=0.6)
    ax.set_title(col)
    ax.legend()
plt.tight_layout()
plt.show()
```
